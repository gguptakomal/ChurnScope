```python
from tensorflow.keras.models import load_model
```

```python
model = load_model('model.h5')
```

```python
import pickle
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)

with open('one_hot_encoder_geo.pkl', 'rb') as file:
    one_hot_encoder_geo = pickle.load(file)

with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
##[input] 
input_data = {
    'CreditScore': 800,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 70000,
    'NumOfProducts': 4,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 90000
}
```

```python
import tensorflow as tf
import numpy as np
import pandas as pd
```

```python
encode_geo = one_hot_encoder_geo.transform([[input_data['Geography']]]).toarray()
encode_geo_df = pd.DataFrame(encode_geo, columns= one_hot_encoder_geo.get_feature_names_out(['Geography']))
encode_geo_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop("Geography", axis = 1), encode_geo_df],axis = 1)
input_df
```

```python
input_scale = scaler.transform(input_df)
```

```python
input_scale
```

```python
##Predict churn
prediction = model.predict(input_scale)
prediction_probability = prediction[0][0]
prediction_probability
```

```python
if prediction_probability > 0.5 : 
    print("Customer will likely churn")
else:
    print("Customer will likely not churn")
```
